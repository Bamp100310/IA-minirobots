Punto 4: Agente de Mantenimiento

```python
import json
from datetime import datetime

class AgenteMantenimiento:
    def __init__(self):
        # Base de datos simulada en memoria
        self.bd_almacen = {
            "BALERO-001": {"cantidad": 5, "ubicacion": "Estante A2"},
            "SENSOR-X10": {"cantidad": 0, "ubicacion": "Estante B1"}
        }
        self.historial_fallas = []
        self.ordenes_trabajo = []

    def buscar_repuesto(self, codigo_repuesto: str) -> bool:
        """Verifica si el repuesto existe en el catálogo."""
        existe = codigo_repuesto in self.bd_almacen
        print(f"[*] Búsqueda de {codigo_repuesto}: {'Encontrado' if existe else 'No encontrado'}")
        return existe

    def consultar_almacen(self, codigo_repuesto: str) -> dict:
        """Retorna la cantidad disponible y ubicación."""
        datos = self.bd_almacen.get(codigo_repuesto, {"cantidad": 0, "ubicacion": "N/A"})
        print(f"[*] Stock de {codigo_repuesto}: {datos['cantidad']} unidades disponibles.")
        return datos

    def generar_orden_trabajo(self, equipo: str, falla: str) -> dict:
        """Crea la orden de trabajo para el equipo afectado."""
        orden = {
            "id_orden": f"OT-{len(self.ordenes_trabajo) + 100}",
            "equipo": equipo,
            "falla": falla,
            "estado": "Pendiente de ejecución",
            "fecha_creacion": datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        }
        self.ordenes_trabajo.append(orden)
        print(f"[*] Orden generada: {orden['id_orden']} -> Equipo: {equipo} | Falla: {falla}")
        return orden

    def actualizar_historial(self, equipo: str, falla: str):
        """Añade el registro al log histórico."""
        registro = {
            "equipo": equipo,
            "falla": falla,
            "fecha": datetime.now().strftime("%Y-%m-%d")
        }
        self.historial_fallas.append(registro)
        print(f"[*] Historial de fallas actualizado para {equipo}.")

    def ejecutar_flujo_agente(self, equipo: str, falla: str, repuesto_necesario: str):
        """Orquesta las herramientas del agente."""
        if self.buscar_repuesto(repuesto_necesario):
            stock = self.consultar_almacen(repuesto_necesario)
            if stock["cantidad"] > 0:
                self.generar_orden_trabajo(equipo, falla)
                self.actualizar_historial(equipo, falla)
                # Descontar stock
                self.bd_almacen[repuesto_necesario]["cantidad"] -= 1
                return "\n[+] Flujo del agente completado exitosamente."
            else:
                return "\n[-] Error: El repuesto no tiene stock."
        return "\n[-] Error: Repuesto no válido."

# Celda de ejecución
agente = AgenteMantenimiento()
resultado = agente.ejecutar_flujo_agente(
    equipo="Bomba Hidráulica 03",
    falla="Vibración excesiva en el eje",
    repuesto_necesario="BALERO-001"
)
print(resultado)
```
