Punto 5: LLM en un microcontrolador (ESP32)

Sí es posible utilizar un LLM en una aplicación con un ESP32, pero únicamente consumiéndolo mediante una API en la nube, no ejecutándolo localmente. La arquitectura del microcontrolador impone las siguientes restricciones técnicas:
*   Memoria insuficiente (RAM/Flash): Un ESP32 típico cuenta con ~520 KB de SRAM y unos pocos MB de memoria Flash. Los LLM (incluso los modelos fundacionales más cuantizados) requieren gigabytes o cientos de megabytes de RAM para cargar los tensores y procesar los tokens de contexto.   
*   Poder de cómputo limitante: El ESP32 no posee GPU, NPU, ni aceleradores de hardware para multiplicaciones de matrices a gran escala. La arquitectura de Transformadores (Multi-head attention, redes FeedForward) requiere un nivel de procesamiento paralelo que el ESP32 no soporta.   
*   Sobrecarga de conexión y latencia: Al depender de conectividad WiFi/Bluetooth para acceder a la API de OpenAI, Claude o un servidor propio con Ollama, el sistema pierde capacidad operativa offline y añade latencia de red en cada inferencia.
*   Consumo de energía: Mantener el radio WiFi encendido y sostener conexiones seguras (TLS/SSL) para enviar prompts y recibir respuestas drena rápidamente la batería en dispositivos IoT remotos.
*   Costos de procesamiento de red: Cifrar la comunicación mediante HTTPS para enviar telemetría a la API consume ciclos de reloj importantes del microcontrolador, dejando menos tiempo para tareas de hardware en tiempo real (manejo de motores, lectura de ADC).
